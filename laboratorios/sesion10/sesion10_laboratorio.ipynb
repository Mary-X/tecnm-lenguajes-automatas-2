````python
%%writefile artesano_sesion10_c.c
/* artesano_sesion10_c.c - Fase 1: El Artesano (sin IA)
 * Mini-evaluador de expresiones (Hito 1).
 *
 * Implementa lo que declara api_sesion10.h.
 * El harness compila este archivo junto con el suyo: no escribas main().
 */
#include "api_sesion10.h"
#include <stdio.h>
#include <string.h>
#include <ctype.h>
#include <stdlib.h>

typedef struct { const char *s; int error; } Parser;

static void saltar(Parser *p) {
    while (*p->s == ' ') p->s++;
}

long expr(Parser *p);

long primaria(Parser *p) {
    saltar(p);
    if (p->error) return 0;

    if (*p->s == '(') {
        p->s++; /* consumir '(' */
        long v = expr(p);
        if (p->error) return 0;
        saltar(p);
        if (*p->s == ')') {
            p->s++; /* consumir ')' */
            return v;
        } else {
            p->error = 2; /* error de sintaxis: falta parentesis de cierre */
            return 0;
        }
    }

    if (isdigit((unsigned char)*p->s)) {
        long v = 0;
        while (isdigit((unsigned char)*p->s)) {
            v = v * 10 + (*p->s - '0');
            p->s++;
        }
        return v;
    }

    p->error = 2; /* error de sintaxis: se esperaba numero o '(' */
    return 0;
}

long termino(Parser *p) {
    long v = primaria(p);
    if (p->error) return 0;
    while (1) {
        saltar(p);
        if (*p->s == '*') {
            p->s++;
            long der = primaria(p);
            if (p->error) return 0;
            v *= der;
        } else if (*p->s == '/') {
            p->s++;
            long der = primaria(p);
            if (p->error) return 0;
            if (der == 0) {
                p->error = 1; /* division entre cero */
                return 0;
            }
            v /= der;
        } else {
            break;
        }
    }
    return v;
}

long expr(Parser *p) {
    long v = termino(p);
    if (p->error) return 0;
    while (1) {
        saltar(p);
        if (*p->s == '+') {
            p->s++;
            long der = termino(p);
            if (p->error) return 0;
            v += der;
        } else if (*p->s == '-') {
            p->s++;
            long der = termino(p);
            if (p->error) return 0;
            v -= der;
        } else {
            break;
        }
    }
    return v;
}

long evaluar_expresion(const char *fuente, int *error) {
    static int sin_error;
    if (error == NULL) error = &sin_error;
    *error = 0;
    if (fuente == NULL) { *error = 2; return 0; }
    Parser p = { fuente, 0 };
    long v = expr(&p);
    saltar(&p);
    if (p.error == 0 && *p.s != '\0') p.error = 2;
    *error = p.error;
    return p.error != 0 ? 0 : v;
}
````

  - No utilices variables globales que dejen estado entre ejecuciones; utiliza la estructura interna provista (`Parser`).

````python
%%writefile ia_sesion10_c.c
/* ia_sesion10_c.c - Fase 2: El Cientifico
 *
 * Codigo generado bajo contrato C-R-E-O.
 * Sigue la firma declarada en api_sesion10.h y no contiene main().
 */
#include "api_sesion10.h"
#include <stdio.h>
#include <stdlib.h>
#include <ctype.h>

typedef struct {
    const char *s;
    int error;
} Parser;

static void saltar_espacios(Parser *p) {
    while (*p->s == ' ') {
        p->s++;
    }
}

long expr(Parser *p);

long primaria(Parser *p) {
    saltar_espacios(p);
    if (p->error) return 0;

    if (*p->s == '(') {
        p->s++; /* consumir '(' */
        long v = expr(p);
        if (p->error) return 0;
        saltar_espacios(p);
        if (*p->s == ')') {
            p->s++; /* consumir ')' */
            return v;
        } else {
            p->error = 2; /* Error de sintaxis: parentesis no cerrado */
            return 0;
        }
    }

    if (isdigit((unsigned char)*p->s)) {
        long v = 0;
        while (isdigit((unsigned char)*p->s)) {
            v = v * 10 + (*p->s - '0');
            p->s++;
        }
        return v;
    }

    p->error = 2; /* Error de sintaxis: se esperaba un operando */
    return 0;
}

long termino(Parser *p) {
    long v = primaria(p);
    if (p->error) return 0;
    while (1) {
        saltar_espacios(p);
        if (*p->s == '*') {
            p->s++;
            long der = primaria(p);
            if (p->error) return 0;
            v *= der;
        } else if (*p->s == '/') {
            p->s++;
            long der = primaria(p);
            if (p->error) return 0;
            if (der == 0) {
                p->error = 1; /* Division entre cero */
                return 0;
            }
            v /= der;
        } else {
            break;
        }
    }
    return v;
}

long expr(Parser *p) {
    long v = termino(p);
    if (p->error) return 0;
    while (1) {
        saltar_espacios(p);
        if (*p->s == '+') {
            p->s++;
            long der = termino(p);
            if (p->error) return 0;
            v += der;
        } else if (*p->s == '-') {
            p->s++;
            long der = termino(p);
            if (p->error) return 0;
            v -= der;
        } else {
            break;
        }
    }
    return v;
}

long evaluar_expresion(const char *fuente, int *error) {
    static int temporal_error;
    if (error == NULL) error = &temporal_error;
    *error = 0;
    if (fuente == NULL) {
        *error = 2;
        return 0;
    }

    Parser p = {fuente, 0};
    long v = expr(&p);
    saltar_espacios(&p);
    if (p.error == 0 && *p.s != '\0') {
        p.error = 2; /* Caracteres adicionales no analizados */
    }
    *error = p.error;
    return p.error != 0 ? 0 : v;
}
````
